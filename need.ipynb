```python
from datasets import load_dataset

# Load all data (train + validation + test)
ds = load_dataset("roskoN/dailydialog", split="train+validation+test")

# Write utterances to file
with open("DailyDialog.txt", "w", encoding="utf-8") as f:
    for ex in ds:
        utterances = ex.get("utterances")
        if utterances and isinstance(utterances, list):
            for line in utterances:
                f.write(line.strip() + "\n")
            f.write("\n")  # Blank line between conversations

print("✅ Saved DailyDialog.txt successfully!")
```
